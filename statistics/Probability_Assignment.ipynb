## Question 1

Kalpa's paid membership renewal benchmark is **72%**. Following a change in onboarding, a random sample of **600 members who have reached their renewal date** contains **408 renewals**.

The membership head suspects that the renewal rate has **fallen below the benchmark**. Test this suspicion at the **5% significance level**.

Also report the difference between the observed renewal rate and the benchmark in **percentage points**. Explain why members who have not yet reached their renewal date should not be included in this sample.

```python
n=600
renewals=408
p_0=0.72
p_1=(renewals/n)
se=sqrt((p_0)*(1-p_0)/n)
z=(p_1-p_0)/se
print(z)
```

```python
# Left Tailed
p_val=norm.cdf(z)
print(p_val)
```

Since p_val is less than 0.05 we reject Null Hypothesis and that there is evidence that the renewal rate has fallen below the benchmark rate.
We should not include members who have not yet reached their renewal rate because thier renewal date has not been reached yet and therefore it would bias the rate downwards when it should not.

## Question 4

A filling machine is calibrated to a population mean fill volume of **500 mL**. Its known population standard deviation is **8 mL**. A random sample of **100 bottles** has a mean volume of **501.7 mL**.

Quality control wants to know whether the population mean fill volume has **changed in either direction**, rather than testing only for overfilling. Use a **5% significance level**.

Give both critical z values. Explain why choosing a right-tailed test after noticing that the sample mean is above 500 mL would change the original question.

```python
mu=500
x_bar=501.7
n=100
sd=8
se=sd/sqrt(n)
```

The critical Z values are +1.96 and -1.96.
Two tailed since it is either direction the p_value is less than 0.05.


Hence, there is enough evidence to say that the rate is above 65%.
