```python
# Question 1: one-proportion z test
p0 = 0.72
n = 600
renewals = 408
alpha = 0.05

p_hat = renewals / n
se = math.sqrt(p0 * (1 - p0) / n)
z = (p_hat - p0) / se
p_value = stats.norm.cdf(z)  # Left-tailed p-value
difference_pp = (p_hat - p0) * 100

print("Hypotheses: H0: p = 0.72; HA: p < 0.72, where p is the population renewal proportion among members who have reached their renewal date.")
print("Tail: Left-tailed, because the suspicion is that the renewal rate has fallen below the benchmark.")
# print(f"Assumptions: Random, independent observations and a population much larger than the sample are specified. Under H0, expected renewals = {n * p0:.0f} and expected non-renewals = {n * (1 - p0):.0f}; both are at least 10, so the normal approximation is appropriate.")
print(f"Observed proportion: {p_hat:.2f} ")
print(f"Standard error: {se:.2f}")
print(f"z statistic: {z:.2f}")
print(f"Left-tailed p-value: {p_value:.2f}")

if p_value < alpha:
    print(f"Decision at alpha = {alpha:.2f}: Reject H0.")
    # print("Conclusion: There is statistically significant evidence at the 5% level that the population renewal rate among members who have reached their renewal date is below 72%. This does not establish the exact population rate.")
else:
    print(f"Decision at alpha = {alpha:.2f}: Do not reject H0.")
    # print("Conclusion: There is insufficient evidence at the 5% level that the population renewal rate among members who have reached their renewal date is below 72%. This does not show that the null hypothesis is true.")

```

```python
# Question 2: one-sample z test for a population mean
mu0 = 42.0
sigma = 12.0  # Known population standard deviation
n = 144
x_ = 39.8
alpha = 0.01

se = sigma / math.sqrt(n)
z = (x_ - mu0) / se
p_value = stats.norm.cdf(z)  # Left-tailed p-value

print("Hypotheses: H0: mu = 42 minutes; HA: mu < 42 minutes, where mu is the population mean order-packing time.")
print("Tail: Left-tailed, because the claim is that the new layout reduced the population mean packing time.")
print("Assumptions: The population standard deviation is known (12 minutes); the sample is random, observations are independent, and the population is much larger than the sample. The z test is therefore appropriate under the stated assumptions.")
print(f"Standard error: {se:.3f} minutes")
print(f"z statistic: {z:.3f}")
print(f"Left-tailed p-value: {p_value:.4f}")

if p_value < alpha:
    print(f"Decision at alpha = {alpha:.2f}: Reject H0.")
    # print("Conclusion: There is statistically significant evidence at the 1% level that the new layout reduced the population mean order-packing time.")
else:
    print(f"Decision at alpha = {alpha:.2f}: Do not reject H0.")
    # print("Conclusion: There is insufficient evidence at the 1% level that the new layout reduced the population mean order-packing time. This does not prove that the population mean is still 42 minutes or that the layout had no effect.")

# print("Assessment of the manager's statement: The sample mean (39.8 minutes) is lower than 42 minutes, but a lower sample average alone does not prove an improvement. At the 1% significance level, the p-value does not provide sufficient evidence to conclude that the population mean packing time decreased.")
```

```python
# Question 3: one-proportion z test
p0, n, defective, alpha = 0.02, 1500, 47, 0.01
p_hat = defective / n
se = math.sqrt(p0 * (1 - p0) / n)
z = (p_hat - p0) / se
p_value = 1 - stats.norm.cdf(z)  # Right-tailed test
expected_defective = n * p0

print("Hypotheses: H0: p = 0.02; HA: p > 0.02, where p is the population defective-item proportion.")
print("Tail: Right-tailed, because escalation requires evidence that the defect rate exceeds 2%.")
print(f"Assumptions: Random, independent observations and a much larger population are specified. Under H0, expected defectives = {expected_defective:.0f} and expected non-defectives = {n * (1-p0):.0f}; both are at least 10, supporting the normal approximation.")
print(f"Observed proportion: {p_hat:.6f}; standard error: {se:.8f}; z statistic: {z:.3f}; right-tailed p-value: {p_value:.4f}.")
print(f"Expected number defective under H0: {expected_defective:.0f}.")
print(f"Decision at alpha = {alpha:.2f}: {'Reject H0.' if p_value < alpha else 'Do not reject H0.'}")
# print("Conclusion: There is evidence at the 1% level that the population defect rate exceeds 2%; escalation is supported." if p_value < alpha else "Conclusion: There is insufficient evidence at the 1% level that the population defect rate exceeds 2%; this does not prove the rate is exactly 2%.")
# print("Comparing 47 with the null expected count of 30 alone is not enough: random samples vary. The z statistic standardizes the difference using the null standard error, and the p-value measures how surprising a result at least this high would be under H0.")
```

```python
# Question 4: two-sided one-sample z test for a mean


mu0, sigma, n, sample_mean, alpha = 500.0, 8.0, 100, 501.7, 0.05
se = sigma / math.sqrt(n)
z = (sample_mean - mu0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))
critical_z = stats.norm.cdf(1 - alpha / 2)

print("Hypotheses: H0: mu = 500 mL; HA: mu != 500 mL, where mu is the population mean fill volume.")
print("Tail: Two-tailed, because quality control asks whether the mean has changed in either direction.")
print("Assumptions: Fill volumes are normally distributed, the population standard deviation is known (8 mL), and the sample is random and independent; the z test is appropriate.")
print(f"Standard error: {se:.3f} mL; z statistic: {z:.3f}; two-tailed p-value: {p_value:.4f}.")
print(f"Critical z values at alpha = {alpha:.2f}: {-critical_z:.3f} and {critical_z:.3f}.")
print(f"Decision: {'Reject H0.' if p_value < alpha else 'Do not reject H0.'}")
# print("Conclusion: There is evidence at the 5% level that the population mean fill volume differs from 500 mL." if p_value < alpha else "Conclusion: There is insufficient evidence at the 5% level that the population mean fill volume differs from 500 mL; this does not establish that the mean is 500 mL.")
# print("Choosing a right-tailed test after seeing the sample mean would change the pre-stated question from a change in either direction to an increase only, and would use the observed data to select the alternative.")
```

```python
# Question 5: one-proportion z test plus separate business threshold

p0, n, completed, alpha = 0.65, 800, 544, 0.05
p_hat = completed / n
se = math.sqrt(p0 * (1 - p0) / n)
z = (p_hat - p0) / se
p_value = 1 - stats.norm.cdf(z)  # Right-tailed test
increase_pp = (p_hat - p0) * 100

print("Hypotheses: H0: p = 0.65; HA: p > 0.65, where p is the population course-completion proportion for learners with the full completion window.")
print("Tail: Right-tailed, because the claim is that completion is higher than 65%.")
print(f"Assumptions: The sample is random and independent, the population is much larger than the sample, and under H0 the expected completed and not-completed counts are {n*p0:.0f} and {n*(1-p0):.0f}, both at least 10.")
print(f"Observed proportion: {p_hat:.3f}; standard error: {se:.8f}; z statistic: {z:.3f}; right-tailed p-value: {p_value:.4f}.")
stat_sig = p_value < alpha
business_met = increase_pp >= 5
print(f"Statistical criterion at alpha = {alpha:.2f}: {'Met; reject H0.' if stat_sig else 'Not met; do not reject H0.'}")
print(f"Observed increase: {increase_pp:.1f} percentage points. Business criterion (at least 5 percentage points): {'Met.' if business_met else 'Not met.'}")
print("Conclusion: There is statistically significant evidence that the completion rate exceeds 65%. The observed improvement is not substantial by the stated business rule because it is less than 5 percentage points." if stat_sig and not business_met else "Conclusion: The statistical test and business threshold are assessed separately as reported above.")
```

```python
# Question 6: two-sided one-sample z test for a mean

mu0, sigma, n, sample_mean, alpha = 18.0, 6.0, 81, 19.4, 0.01
se = sigma / math.sqrt(n)
z = (sample_mean - mu0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("Hypotheses: H0: mu = 18 points per day; HA: mu != 18 points per day, where mu is the population mean productivity score.")
print("Tail: Two-tailed, because the original review asks whether the mean differs from 18 in either direction.")
print("Assumptions: The population standard deviation is known (6 points), and the 81 employee-day records are independent and randomly sampled from a much larger population; the z test is appropriate.")
print(f"Standard error: {se:.6f} points; z statistic: {z:.3f}; two-tailed p-value: {p_value:.4f}.")
print(f"Decision at alpha = {alpha:.2f}: {'Reject H0.' if p_value < alpha else 'Do not reject H0.'}")
print("Conclusion: There is evidence at the 1% level that the population mean productivity differs from 18 points per day." if p_value < alpha else "Conclusion: There is insufficient evidence at the 1% level that the population mean productivity differs from 18 points per day; this does not establish that the mean is 18.")
print("The alternative must reflect the original two-direction review question. Choosing a right-tailed alternative after observing a higher sample mean changes the question and biases the test toward the observed direction.")
```

```python
# Question 7: one-proportion z test and two ways to express the reduction
import math
from statistics import NormalDist

p0, n, repeats, alpha = 0.12, 1250, 127, 0.05
p_hat = repeats / n
se = math.sqrt(p0 * (1 - p0) / n)
z = (p_hat - p0) / se
p_value = stats.norm.cdf(z)  # Left-tailed test
reduction_pp = (p0 - p_hat) * 100
reduction_relative_pct = (p0 - p_hat) / p0 * 100

print("Hypotheses: H0: p = 0.12; HA: p < 0.12, where p is the population seven-day repeat-contact proportion.")
print("Tail: Left-tailed, because the question asks whether the repeat-contact rate is below 12%.")
print(f"Assumptions: The customers are randomly sampled and independent, all completed the observation window, the population is much larger than the sample, and null expected repeat/non-repeat counts ({n*p0:.0f} and {n*(1-p0):.0f}) are both at least 10.")
print(f"Observed proportion: {p_hat:.4f}; standard error: {se:.8f}; z statistic: {z:.3f}; left-tailed p-value: {p_value:.4f}.")
print(f"Decision at alpha = {alpha:.2f}: {'Reject H0.' if p_value < alpha else 'Do not reject H0.'}")
print("Conclusion: There is evidence at the 5% level that the population repeat-contact rate is below 12%." if p_value < alpha else "Conclusion: There is insufficient evidence at the 5% level that the population repeat-contact rate is below 12%; this does not prove it is 12%.")
print(f"Observed reduction: {reduction_pp:.2f} percentage points, or {reduction_relative_pct:.2f}% of the 12% benchmark.")
print("The percentage-point reduction is the absolute difference between rates; the relative percentage expresses that difference as a fraction of the benchmark.")
```

```python
# Question 8: one-sample z test for mean runtime plus adoption threshold
import math
from statistics import NormalDist

mu0, sigma, n, sample_mean, alpha = 8.0, 2.5, 225, 7.72, 0.05
se = sigma / math.sqrt(n)
z = (sample_mean - mu0) / se
p_value = stats.norm.cdf(z)  # Left-tailed test
saving = mu0 - sample_mean

print("Hypotheses: H0: mu = 8 seconds; HA: mu < 8 seconds, where mu is the population mean query runtime.")
print("Tail: Left-tailed, because the question asks whether runtime decreased.")
print("Assumptions: The population standard deviation is known (2.5 seconds); executions are randomly sampled and independent, and the population is much larger than the sample. Thus the z test is appropriate.")
print(f"Standard error: {se:.6f} seconds; z statistic: {z:.3f}; left-tailed p-value: {p_value:.4f}.")
print(f"Decision at alpha = {alpha:.2f}: {'Reject H0.' if p_value < alpha else 'Do not reject H0.'}")
print("Conclusion: There is statistically significant evidence at the 5% level that the population mean runtime decreased." if p_value < alpha else "Conclusion: There is insufficient evidence at the 5% level that the population mean runtime decreased; this does not prove no decrease occurred.")
print(f"Estimated saving per execution: {saving:.2f} seconds.")
print(f"Adoption requirement (at least 0.5 seconds observed saving): {'Met.' if saving >= 0.5 else 'Not met.'} This operational threshold is separate from the hypothesis-test result.")
```

```python
# Question 9: two-sided one-proportion z test plus launch-policy threshold
import math
from statistics import NormalDist

p0, n, purchases, alpha = 0.40, 900, 390, 0.05
p_hat = purchases / n
se = math.sqrt(p0 * (1 - p0) / n)
z = (p_hat - p0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))
increase_pp = (p_hat - p0) * 100

print("Hypotheses: H0: p = 0.40; HA: p != 0.40, where p is the population purchase-completion proportion.")
print("Tail: Two-tailed, because the research question asks whether the rate differs in either direction.")
print(f"Assumptions: Checkout sessions are randomly sampled and independent, the population is much larger than the sample, and under H0 expected completed and uncompleted counts ({n*p0:.0f} and {n*(1-p0):.0f}) are both at least 10.")
print(f"Observed proportion: {p_hat:.4f}; standard error: {se:.8f}; z statistic: {z:.3f}; two-tailed p-value: {p_value:.4f}.")
print(f"Statistical decision at alpha = {alpha:.2f}: {'Reject H0.' if p_value < alpha else 'Do not reject H0.'}")
print("Conclusion: There is evidence at the 5% level that the population completion rate differs from 40%." if p_value < alpha else "Conclusion: There is insufficient evidence at the 5% level that the population completion rate differs from 40%; this does not establish that it is 40%.")
print(f"Observed increase: {increase_pp:.2f} percentage points. Launch-policy requirement (increase of at least 4 percentage points): {'Met.' if increase_pp >= 4 else 'Not met.'}")
print("The statistical test and launch policy do not lead to the same conclusion: the test detects a difference, while the observed increase falls short of the 4-percentage-point launch requirement." if p_value < alpha and increase_pp < 4 else "The statistical test and launch-policy result are reported separately above.")
```

```python
# Question 10: two-sided one-sample z test, decisions at two significance levels
from statistics import NormalDist

mu0, sigma, n, sample_mean = 100.0, 15.0, 196, 102.2
se = sigma / math.sqrt(n)
z = (sample_mean - mu0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("Hypotheses: H0: mu = 100 points; HA: mu != 100 points, where mu is the population mean service-quality index.")
print("Tail: Two-tailed, because the question asks whether the mean changed in either direction.")
print("Assumptions: The population standard deviation is known (15 points); the sample is random and independent, and the population is much larger than the sample, so a z test is appropriate.")
print(f"Standard error: {se:.8f} points; z statistic: {z:.3f}; two-tailed p-value: {p_value:.4f}.")
for alpha in (0.05, 0.01):
    print(f"Decision at alpha = {alpha:.2f}: {'Reject H0.' if p_value < alpha else 'Do not reject H0.'}")
print("At 5%, there is evidence that the population mean index changed in either direction. At 1%, there is insufficient evidence of a change; this does not establish that the mean is 100.")
print("The p-value is compared with each significance cutoff. It is below 0.05 but above 0.01, so the decisions differ even though the sample and test statistic are unchanged.")
```
