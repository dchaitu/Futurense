```python
# Question 1
p0 = 0.72
n = 600
renewals = 408
alpha = 0.05

p_cap = renewals / n
se = math.sqrt(p0 * (1 - p0) / n)
z = (p_cap - p0) / se
p_value = stats.norm.cdf(z)
observed_diff = (p_cap - p0)*100

# Left-tailed p-value
print("z value is ",z)
print("p value is ",p_value)

h0 = "benchmark = 72%"
ha = "benchmark < 72%"

if p_value < alpha:
    print(f"Reject H0.",h0)
else:
    print(f"Do not reject H0.",h0)

print(f"Observed renewal rate difference is {observed_diff:.3f}")

```

```python
# 19
alpha = 0.05
n=1000
p_1 = 37/n
p_0 = 0.03
se=math.sqrt(p_0*(1-p_0)/n)
z=(p_1-p_0)/se
p_val = 1- stats.norm.cdf(z)
H0 = "p=3%"
Ha = "p > 3%"
# Right tailed test
print("z value is ",z)
print("The p value is ",p_val)
if p_val<alpha:
    print("Reject H0")
else:
    print("Failed to reject H0")

z_critc = stats.norm.ppf(1-alpha)
print("z critic value is ",z_critc)
min_num_of_records = (z_critc*se + p_0)*n
print("Minimum number of records is ",math.ceil(min_num_of_records))
```
