```python
import numpy as np
import pandas as pd
```

```python
c = pd.read_csv('covid_19_confirmed_v1.csv')
r = pd.read_csv('covid_19_recovered_v1.csv')
d = pd.read_csv('covid_19_deaths_v1.csv',skiprows=1)
```

```python
c_trans=pd.melt(c,id_vars=['Province/State','Country/Region','Lat','Long'],var_name='Date',value_name='Cases')
```

```python
c_trans
```

```python
r_trans=pd.melt(r,id_vars=['Province/State','Country/Region','Lat','Long'],var_name='Date',value_name='recovered_cases')
r_trans
```

```python
d_trans=pd.melt(d,id_vars=['Province/State','Country/Region','Lat','Long'],var_name='Date',value_name='deaths')
d_trans
```

```python
c_trans['Date'] = pd.to_datetime(c_trans.Date)
```

```python
r_trans['Date'] = pd.to_datetime(r_trans.Date)
```

```python
d_trans['Date'] = pd.to_datetime(d_trans.Date)
```

```python
import seaborn as sns

sns.lineplot(data=c_trans,x='Date',y="Cases",estimator='sum')
```

```python
import matplotlib.pyplot as plt

plt.bar(c_trans.Date,c_trans.Cases)
```
