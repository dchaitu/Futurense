- Heights of adult men in a city have μ = 170 cm and σ = 8 cm. For a random sample of n = 64 men, what is the standard error of the sample mean?

- If you want to cut the standard error in half, by what factor must you increase n?

```python
u= 170
sd = 8
n=64
se = sd/math.sqrt(n)
print(se)
```

```python
u=50
sd=12
x_ = 52
n=36
se = sd/math.sqrt(n)
print(se)
z = (x_-u)/se
print(z)
# right side curve
p = 1- stats.norm.cdf(z)
print(p)
```

- A fair coin is flipped 400 times. Use the CLT to approximate P(190 ≤ heads ≤ 210). Then try the same problem with a continuity correction and compare.
-

```python
n = 400
p = 0.5
x_ = n*p
sd = math.sqrt(n*p*(1-p))
print("sd ",sd)
se = math.sqrt(p*(1-p)/n)
print("se",se)
z_min = ((190/n-p)/se)
z_max = ((210/n-p)/se)
print("z_min",z_min)
print("z_max",z_max)
p_min = stats.norm.cdf(z_min)
p_max = stats.norm.cdf(z_max)
print("p_min",p_min)
print("p_max",p_max)
prob = (p_max-p_min)
print("prob ",prob)

```

- Delivery times have mean 30 min and σ = 10 min, but the distribution is right-skewed. A courier makes 40 deliveries per shift. What’s the probability their total time exceeds 1,300 minutes?

```python
u = 30
sd = 10
# Right tailed curve
n = 40
x_ = 1300/n

se = sd/math.sqrt(n)
z = (x_-u)/se
print("se ", se)
print("z ",z)
p = 1-stats.norm.cdf(z)
print("p ",p)
```

```python
import pandas as pd
plus = pd.DataFrame({
"member": ["M1", "M1", "M1", "M2", "M2"],
"month": ["Jun", "Jun", "Jul", "Jun", "Jun"],
"amount": [3000, 1000, 2400, 1800, 600]})
wide = plus.pivot_table(index="member", columns="month", values="amount",aggfunc="mean")
display(wide)
long = wide.reset_index().melt(id_vars="member", value_name="amount")
display(long)
print(wide.loc["M1", "Jun"], len(long), long["amount"].sum())
```

```python
import pandas as pd

members = pd.DataFrame({
    "city": ["Hyderabad", "Hyderabad", "Pune", "Pune"],
    "annual_fee": [1000, 1500, 2000, 2500]
})
members
```

```python
def fee_difference(group):
    return group["annual_fee"] - group["annual_fee"].mean()
```

```python
result = members.groupby("city").apply(fee_difference)
print(result)
```

```python
result = members.groupby("city", group_keys=False).apply(fee_difference)
print(result)
```
