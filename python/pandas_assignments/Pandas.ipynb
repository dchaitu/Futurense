```python
import pandas as pd
import numpy as np

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106,
                 107, 108, 109, 110, 111, 112],
    'customer_id': ['C1', 'C2', 'C1', 'C3', 'C2', 'C4',
                    'C3', 'C1', 'C4', 'C2', 'C5', 'C5'],
    'city': ['Hyd', 'Delhi', 'Hyd', 'Mumbai', 'Delhi', 'Hyd',
             'Mumbai', 'Hyd', 'Delhi', 'Delhi', 'Hyd', 'Mumbai'],
    'category': ['Electronics', 'Fashion', 'Fashion',
                 'Electronics', 'Electronics', 'Fashion',
                 'Fashion', 'Electronics', 'Fashion',
                 'Electronics', 'Fashion', 'Electronics'],
    'amount': [5000, 2000, 1500, 7000, 3000, 2500,
               4000, 6000, 3500, 4500, 2000, 5500],
    'discount': [500, 200, 0, 700, 300, 250,
                 400, 600, 0, 450, 200, 550],
    'quantity': [2, 1, 3, 2, 1, 2, 1, 3, 2, 1, 2, 2],
    'status': ['Completed', 'Completed', 'Cancelled',
               'Completed', 'Completed', 'Completed',
               'Cancelled', 'Completed', 'Completed',
               'Completed', 'Completed', 'Completed']
})
```

```python
# Calculate the total order amount for each city.
#
# Expected output: One row per city.

orders.groupby('city').agg(total_amount = ('amount', 'sum'))
```

```python
# Problem 2. Multiple aggregations
#
# For each city, calculate:
# - Total order amount.
# - Average order amount.
# - Maximum order amount.
# - Number of orders.

df = orders.groupby('city').agg(total_amount = ('amount', 'sum'),
                           avg_amount = ('amount', 'mean'),
                           max_amount = ('amount', 'max'),
                           order_count = ( 'order_id', 'count'))
df
```

```python
# City average for every order
#
# Create a new column called city_avg_amount containing the average order amount for the corresponding city.
# Important: The original DataFrame must retain all 12 rows.

# orders.groupby('city')['amount'].transform('mean')
orders['city_avg_amount'] = (
    orders.groupby('city')['amount'].transform('mean')
)
orders
```

```python
# Problem 4. Customer spending
#
# For each customer, calculate the total amount spent on completed orders.
completed_orders = orders[
    orders['status'] == 'Completed'
]
completed_orders.groupby('customer_id').agg(total_amount=('amount','sum'))
```

```python
# Problem 5. Custom group calculation
#
# For each city, calculate:
# \[
# \text{Sales Range} = \text{Maximum Amount} - \text{Minimum Amount}
# \]
#
# Use apply() to solve this problem.

def get_sales_range(df):
    return df['amount'].max()-df['amount'].min()

orders.groupby('city').apply(get_sales_range)
```

```python
result = orders.groupby('city').agg(
    sales_range=('amount', lambda x: x.max() - x.min())
)
result
```

```python
orders.groupby('city').agg('size')
```

```python
orders.groupby('city').agg('count')
```

```python
# Problem 7. Filter customers
#
# Keep only customers whose total completed-order amount is greater than ₹7,000.
#
# Your output must contain the original order records of qualifying customers, not just their aggregated totals.
completed_orders = orders[(orders['status'] == 'Completed')&(orders['total_amount'] > 7000)]
completed_orders['total_amount'] = completed_orders.groupby('customer_id')['amount'].transform('sum')

completed_orders
```

```python
import pandas as pd

data = {
    "Student": ["Alice", "Bob"],
    "Math": [85, 90],
    "Science": [92, 80],
    "English": [78, 88],
}

df = pd.DataFrame(data)
print(df)
```

```python
df_melted = pd.melt(
    df,
    id_vars=["Student"],  # Identifier column (stays as a column)
    value_vars=["Math", "Science", "English"],  # Columns to melt into rows
    var_name="Subject",  # New column name for subject headers
    value_name="Score",  # New column name for test scores
)

print(df_melted)
```
