## Task 3: Fix the Missing Values

- Fill the blanks in `Age` with the **median** age. Store the median in `age_median`.
- Fill the blanks in `Embarked` with the **most common** port. Store it in `embarked_mode`.
- Store the number of blanks left in these two columns in `blanks_left`.

As in class, try the fix on a small sample first.

### 3.1 Try It on a Sample of 10 Rows

```python
# 7 rows from the top, plus 3 rows where Age is blank
small = pd.concat([df.head(7), df[df["Age"].isna()].head(3)])

age_median = df["Age"].median()
small["Age"] = small["Age"].mode()

print("Median age:", age_median)
small[["Name", "Age"]]
```

### 3.2 Apply the Fix to All Rows

```python
df["Age"] = df["Age"].fillna(age_median)

# mode() gives the most common value. [0] takes the first one.
embarked_mode = df["Embarked"].mode()[0]
df["Embarked"] = df["Embarked"].fillna(embarked_mode)

# Blanks left in Age and Embarked together
blanks_left = df[["Age", "Embarked"]].isna().sum()[0]

print("Most common port:", embarked_mode, "  Blanks left:", blanks_left)
```

## Task 4: Prepare the Columns

The inputs are given in `feature_cols`. `Sex` and `Embarked` are text.

- Build `X` with text columns turned into 0/1 columns, dropping the first category of each.
- Build `y` from the `Survived` column.
- Store the number of input columns in `n_features`.
- Store the number of male passengers in `n_male`, using the new 0/1 column for sex.

*Hint:* print `X.columns` to find the name of the new column.

```python
# Name, Ticket and Cabin are left out
feature_cols = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]

X = pd.get_dummies(df[feature_cols],  drop_first=True, dtype=int)
y = df["Survived"]

X.columns
```

```python
n_features = len(X.columns)
n_male = X['Sex_male'].sum()

print("Input columns:", n_features, "  Male passengers:", n_male)
```

## Task 5: Split into Train and Test

Use the same settings as in class: 25 percent test rows, `random_state=42`, and keep the share of survivors the same in both parts.

- `n_test`: the number of test rows
- `test_survivors`: the number of survivors in the test rows

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

n_test = len(X_test)
test_survivors = y_test.sum()

print("Test rows:", n_test, "  Survivors in test rows:", test_survivors)
```

## Task 6: Train, Predict and Compare with the Baseline

- Train a `LogisticRegression(max_iter=2000)` on the training rows.
- Store the chance of survival for each test passenger in `scores`.
- Store yes/no predictions at cut-off 0.5 in `y_pred_05`, drawing the line yourself.
- `baseline_accuracy`: the accuracy of a rule that always predicts the most common answer.
- `accuracy_05`: the accuracy of the model at 0.5.

```python
model = LogisticRegression(max_iter=2000)
model.fit(X_train, y_train)

scores = model.predict_proba(X_test)[:, 1]
y_pred_05 = (scores >= 0.5).astype(int)

# The share of the most common answer in the test rows
baseline_accuracy = y_test.value_counts(normalize=True).max()
accuracy_05 = accuracy_score(y_test, y_pred_05)

print("Baseline accuracy:", round(baseline_accuracy, 3))
print("Model accuracy at 0.5:", round(accuracy_05, 3))
```

**Check for Task 6**

```python
check("Task 6", ["baseline_accuracy", "accuracy_05"])
```

## Task 7: The Four Boxes, Precision and Recall

At cut-off 0.5:

- Unpack the four boxes into `tn_05`, `fp_05`, `fn_05`, `tp_05`.
- Compute `precision_05` and `recall_05` **by hand** from the four boxes. Do not use `precision_score` or `recall_score`.

*Remember the order sklearn uses when you unpack.*

```python
cm = confusion_matrix(y_test, y_pred_05)
tn_05, fp_05, fn_05, tp_05 = cm.ravel()

precision_05 = tp_05 / (tp_05 + fp_05)
recall_05 = tp_05/ (tp_05 + fn_05)

print("Caught:", tp_05, "  False alarms:", fp_05, "  Missed:", fn_05, "  Correct no:", tn_05)
print("Precision:", round(precision_05, 3), "  Recall:", round(recall_05, 3))
```

**Check for Task 7**

```python
check("Task 7", ["tp_05", "fp_05", "fn_05", "tn_05", "precision_05", "recall_05"])
```

## Task 8: Move the Cut-off and Decide

The rule for this task:

1. Recall must be **at least 0.75**.
2. A missed survivor costs **5 points**. A false alarm costs **1 point**.
3. Among the cut-offs that meet the recall rule, choose the one with the **lowest cost**.

Compute `recall_04` and `recall_03` at cut-offs 0.4 and 0.3. Compute the cost at all three cut-offs: `cost_05`, `cost_04`, `cost_03`. Then answer:

- `best_cutoff`: your chosen cut-off, as a number (0.5, 0.4 or 0.3)
- `accuracy_change`: did accuracy go `"up"` or `"down"` when the line moved from 0.5 to 0.4? Write the word as text.

### 8.1 Cut-off 0.4

```python
y_pred_04 = (scores >= 0.4).astype(int)
cm = confusion_matrix(y_test, y_pred_04)
tn_04, fp_04, fn_04, tp_04 = cm.ravel()

recall_04 = tp_04 / (tp_04 + fn_04)
accuracy_04 = accuracy_score(y_test, y_pred_04)

print("Recall at 0.4:", round(recall_04, 3), "  Accuracy at 0.4:", round(accuracy_04, 3))
```

### 8.2 Cut-off 0.3

```python
y_pred_03 = (scores >= 0.3).astype(int)
cm = confusion_matrix(y_test, y_pred_03)

tn_03, fp_03, fn_03, tp_03 = cm.ravel()

recall_03 = tp_03/(tp_03+fn_03)
accuracy_03 = accuracy_score(y_test, y_pred_03)
print("Recall at 0.3:", round(recall_03, 3), "  Accuracy at 0.3:", round(accuracy_03, 3))
```

### 8.3 The Cost at Each Cut-off

```python
cost_miss = 5
cost_false_alarm = 1

cost_05 = fn_05 * cost_miss + fp_05 * cost_false_alarm
cost_04 = fn_04 * cost_miss + fp_04 * cost_false_alarm
cost_03 = fn_03 * cost_miss + fp_03 * cost_false_alarm

print("Cost at 0.5:", cost_05, "  Cost at 0.4:", cost_04, "  Cost at 0.3:", cost_03)
```

### 8.4 Your Decision

```python
best_cutoff = 0.4
accuracy_change = "up"
```
